```python

# EXPERIMENT: Data Visualization using Python

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create a sample dataset instead of loading a CSV file
df = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D', 'E'],
    'Values': [3, 7, 5, 4, 8]
})

print("Sample Dataset:")
print(df.head())


# 1. LINE CHART
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, color='blue', label='Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot')
plt.legend()
plt.grid(True)
plt.show()


# 2. BAR CHART
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 5, 4]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color='orange')
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Bar Plot')
plt.show()


# 3. HISTOGRAM
data = np.random.randn(1000)

plt.figure(figsize=(8, 5))
plt.hist(data, bins=20, color='purple', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```
